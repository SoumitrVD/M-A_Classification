```python
financing_population_df = (
    financing_population_df
    .withColumn(
        "classification_key",
        F.concat_ws(
            "_",
            F.col("mna_deal_id").cast("string"),
            F.col("financing_deal_id").cast("string")
        )
    )
)
```

```python
will_reviewed_financings_df = (
    will_financings_df
    .withColumn(
        "will_financing_deal_id",
        F.coalesce(
            F.col("loan_id").cast("string"),
            F.col("bond_id").cast("string")
        )
    )
    .groupBy(
        F.col("deal_id").cast("string").alias("will_mna_deal_id"),
        "will_financing_deal_id"
    )
    .agg(
        F.count("*").alias("will_reviewed_tranche_count"),
        F.collect_set("facility_key").alias("will_facility_keys"),
        F.collect_set("instrument").alias("will_instruments")
    )
)
```

```python
financing_population_df = (
    financing_population_df.alias("p")
    .join(
        will_reviewed_financings_df.alias("w"),
        (
            F.col("p.mna_deal_id").cast("string")
            == F.col("w.will_mna_deal_id")
        )
        & (
            F.col("p.financing_deal_id").cast("string")
            == F.col("w.will_financing_deal_id")
        ),
        how="left"
    )
    .drop(
        "will_mna_deal_id",
        "will_financing_deal_id"
    )
    .withColumn(
        "reviewed_by_will",
        F.col("will_reviewed_tranche_count").isNotNull()
    )
)
```

#Combine the supporting evidence

```python
mna_parties_df = (
    mna_deals_df
    .select(
        "mna_deal_id",

        F.expr("""
            element_at(
                filter(
                    MNADealCompanies,
                    x -> lower(x.CompanyRoleName) = 'target'
                       and x.IsPrimary = true
                ),
                1
            )
        """).alias("target"),

        F.expr("""
            element_at(
                filter(
                    MNADealCompanies,
                    x -> lower(x.CompanyRoleName) = 'acquiror'
                       and x.IsPrimary = true
                ),
                1
            )
        """).alias("acquiror")
    )
    .select(
        "mna_deal_id",

        F.col("target.CompanyId").alias("target_id"),
        F.col("target.CompanyName").alias("target_name"),

        F.col("acquiror.CompanyId").alias("acquiror_id"),
        F.col("acquiror.CompanyName").alias("acquiror_name")
    )
)
```

```python
financing_evidence_df = (
    financing_population_df
    .join(
        mna_parties_df,
        on="mna_deal_id",
        how="left"
    )
)
```

```python
financing_evidence_df = (
    financing_evidence_df

    # Extract the primary Loan borrower
    .withColumn(
        "loan_borrower",
        F.expr("""
            element_at(
                filter(
                    LoanDealCompanies,
                    x -> lower(x.RoleName) = 'borrower'
                ),
                1
            )
        """)
    )

    # Create one standard borrower/issuer ID
    .withColumn(
        "borrower_or_issuer_id",
        F.when(
            F.col("financing_market") == "Loan",
            F.col("loan_borrower.CompanyId")
        )
        .when(
            F.col("financing_market") == "DCM",
            F.col("DCMDeal_IssuerId")
        )
        .when(
            F.col("financing_market") == "ECM",
            F.col("ECMDeal_IssuerId")
        )
    )

    # Create one standard borrower/issuer name
    .withColumn(
        "borrower_or_issuer_name",
        F.when(
            F.col("financing_market") == "Loan",
            F.col("loan_borrower.CompanyName")
        )
        .when(
            F.col("financing_market") == "DCM",
            F.col("DCMDeal_IssuerName")
        )
        .when(
            F.col("financing_market") == "ECM",
            F.col("ECMDeal_IssuerName")
        )
    )

    .drop("loan_borrower")
)
```

```python
financing_evidence_df = (
    financing_evidence_df

    .withColumn(
        "financing_announcement_date",
        F.to_date(
            F.when(
                F.col("financing_market") == "Loan",
                F.col("LoanDeal_AnnouncementDate")
            )
            .when(
                F.col("financing_market") == "DCM",
                F.col("DCMDeal_AnnouncementDate")
            )
            .when(
                F.col("financing_market") == "ECM",
                F.col("ECMDeal_AnnouncementDate")
            )
        )
    )

    .withColumn(
        "financing_pricing_date",
        F.to_date(
            F.when(
                F.col("financing_market") == "DCM",
                F.col("DCMDeal_PricingDate")
            )
            .when(
                F.col("financing_market") == "ECM",
                F.col("ECMDeal_PricingDate")
            )
        )
    )

    .withColumn(
        "financing_settlement_date",
        F.to_date(
            F.when(
                F.col("financing_market") == "DCM",
                F.col("DCMDeal_SettlementDate")
            )
            .when(
                F.col("financing_market") == "ECM",
                F.col("ECMDeal_SettlementDate")
            )
        )
    )

    .withColumn(
        "financing_sign_date",
        F.to_date(
            F.when(
                F.col("financing_market") == "Loan",
                F.col("SignDate")
            )
        )
    )

    .withColumn(
        "financing_funding_date",
        F.to_date(
            F.when(
                F.col("financing_market") == "Loan",
                F.col("FundingDate")
            )
        )
    )

    .withColumn(
        "financing_type",
        F.when(
            F.col("financing_market") == "Loan",
            F.col("LoanDeal_TypeName")
        )
        .when(
            F.col("financing_market") == "DCM",
            F.col("DCMDeal_TypeName")
        )
        .when(
            F.col("financing_market") == "ECM",
            F.col("ECMDeal_TypeName")
        )
    )
)
```

```python
financing_evidence_df = (
    financing_evidence_df

    .withColumn(
        "financing_use_of_proceeds_raw",
        F.when(
            F.col("financing_market") == "DCM",
            F.to_json(F.col("DCMDealUseOfProceeds"))
        )
        .when(
            F.col("financing_market") == "ECM",
            F.to_json(F.col("ECMDealUseOfProceeds"))
        )
    )

    .withColumn(
        "financing_use_of_proceeds_note",
        F.when(
            F.col("financing_market") == "ECM",
            F.col("UseOfProceedsNote")
        )
        .otherwise(F.col("Comments"))
    )
)
```

##Loan tranche evidence

```python
loan_evidence_df = (
    loan_tranches_df

    .withColumn(
        "loan_use_names",
        F.expr("""
            transform(
                LoanDealTranchesUseOfProceeds,
                x -> x.UseName
            )
        """)
    )

    .groupBy(
        "mna_deal_id",
        "financing_deal_id"
    )

    .agg(
        F.countDistinct("tranche_id")
        .alias("loan_tranche_count"),

        F.sort_array(
            F.collect_set("PrimaryInstrumentTypeName")
        ).alias("loan_instrument_types"),

        F.sort_array(
            F.array_distinct(
                F.flatten(
                    F.collect_list("loan_use_names")
                )
            )
        ).alias("loan_use_of_proceeds"),

        F.max(
            F.coalesce(
                F.col("IsBridgeToEquity").cast("int"),
                F.lit(0)
            )
        ).cast("boolean").alias("has_bridge_to_equity"),

        F.max(
            F.coalesce(
                F.col("IsBridgeToHighYield").cast("int"),
                F.lit(0)
            )
        ).cast("boolean").alias("has_bridge_to_high_yield"),

        F.max(
            F.coalesce(
                F.col("IsSecured").cast("int"),
                F.lit(0)
            )
        ).cast("boolean").alias("has_secured_loan_tranche"),

        F.max(
            F.coalesce(
                F.col("IsSubordinated").cast("int"),
                F.lit(0)
            )
        ).cast("boolean").alias("has_subordinated_loan_tranche"),

        F.min(
            F.to_date("MaturityDate")
        ).alias("earliest_loan_maturity_date"),

        F.max(
            F.to_date("MaturityDate")
        ).alias("latest_loan_maturity_date"),

        F.sort_array(
            F.collect_set("CommonStatusName")
        ).alias("loan_tranche_statuses"),

        F.sort_array(
            F.collect_set("Note")
        ).alias("loan_tranche_notes"),

        F.collect_list(
            F.struct(
                "tranche_id",
                "PrimaryInstrumentTypeName",
                "TypeName",
                "MaturityDate",
                "CancelledDate",
                "CancelledFlag",
                "CommonStatusName",
                "IsBridgeToEquity",
                "IsBridgeToHighYield",
                "IsSecured",
                "IsSubordinated",
                "LoanDealTranchesUseOfProceeds",
                "LoanDealTranchesOutstandings",
                "LoanDealTranchesValue",
                "Note"
            )
        ).alias("loan_tranche_details")
    )
)
```

##Bond tranche evidence

```python
bond_evidence_df = (
    bond_tranches_df

    .groupBy(
        "mna_deal_id",
        "financing_deal_id"
    )

    .agg(
        F.countDistinct("tranche_id")
        .alias("bond_tranche_count"),

        F.sort_array(
            F.collect_set("SecurityTypeName")
        ).alias("bond_security_types"),

        F.sort_array(
            F.collect_set("SeniorityName")
        ).alias("bond_seniority_types"),

        F.max(
            F.coalesce(
                F.col("IsHighYield").cast("int"),
                F.lit(0)
            )
        ).cast("boolean").alias("has_high_yield_bond"),

        F.max(
            F.coalesce(
                F.col("IsLevFin").cast("int"),
                F.lit(0)
            )
        ).cast("boolean").alias("has_levfin_bond"),

        F.max(
            F.coalesce(
                F.col("IsSubordinated").cast("int"),
                F.lit(0)
            )
        ).cast("boolean").alias("has_subordinated_bond"),

        F.min(
            F.to_date("MaturityDate")
        ).alias("earliest_bond_maturity_date"),

        F.max(
            F.to_date("MaturityDate")
        ).alias("latest_bond_maturity_date"),

        F.sort_array(
            F.collect_set("Comment")
        ).alias("bond_comments"),

        F.sort_array(
            F.collect_set("RedemptionDetails")
        ).alias("bond_redemption_details"),

        F.collect_list(
            F.struct(
                "tranche_id",
                "IssueTypeName",
                "SecurityTypeName",
                "SeniorityName",
                "MaturityDate",
                "FirstCallDate",
                "TakeOutDate",
                "IsHighYield",
                "IsLevFin",
                "IsSubordinated",
                "Comment",
                "RedemptionDetails",
                "DCMDealTranchesOutstandingAmount",
                "DCMDealTranchesProceeds"
            )
        ).alias("bond_tranche_details")
    )
)
```

##ECM tranche evidence

```python
ecm_evidence_df = (
    ecm_tranches_df

    .groupBy(
        "mna_deal_id",
        "financing_deal_id"
    )

    .agg(
        F.countDistinct("tranche_id")
        .alias("ecm_tranche_count"),

        F.sort_array(
            F.collect_set("SecurityTypeName")
        ).alias("ecm_security_types"),

        F.sort_array(
            F.collect_set("OfferTypeName")
        ).alias("ecm_offer_types"),

        F.sort_array(
            F.collect_set("SubTypeName")
        ).alias("ecm_subtypes"),

        F.sort_array(
            F.collect_set("Note")
        ).alias("ecm_notes"),

        F.collect_list(
            F.struct(
                "tranche_id",
                "SecurityTypeName",
                "OfferTypeName",
                "SubTypeName",
                "MaturityDate",
                "Shares",
                "SharesDescription",
                "CurrencyISOCode",
                "ECMDealTranchesOfferPrice",
                "ECMDealTranchesValue",
                "Note"
            )
        ).alias("ecm_tranche_details")
    )
)
```

```python
financing_evidence_df = (
    financing_evidence_df

    .join(
        loan_evidence_df,
        on=["mna_deal_id", "financing_deal_id"],
        how="left"
    )

    .join(
        bond_evidence_df,
        on=["mna_deal_id", "financing_deal_id"],
        how="left"
    )

    .join(
        ecm_evidence_df,
        on=["mna_deal_id", "financing_deal_id"],
        how="left"
    )
)
```

```python
financing_evidence_df = (
    financing_evidence_df

    .withColumn(
        "loan_use_of_proceeds_text",
        F.concat_ws(
            " | ",
            F.col("loan_use_of_proceeds")
        )
    )

    .withColumn(
        "bond_comments_text",
        F.concat_ws(
            " | ",
            F.col("bond_comments")
        )
    )

    .withColumn(
        "bond_redemption_text",
        F.concat_ws(
            " | ",
            F.col("bond_redemption_details")
        )
    )

    .withColumn(
        "ecm_notes_text",
        F.concat_ws(
            " | ",
            F.col("ecm_notes")
        )
    )

    .withColumn(
        "combined_evidence_text",
        F.concat_ws(
            " | ",
            F.col("mna_funding_comment"),
            F.col("mna_related_deal_comment"),
            F.col("financing_use_of_proceeds_raw"),
            F.col("financing_use_of_proceeds_note"),
            F.col("loan_use_of_proceeds_text"),
            F.col("bond_comments_text"),
            F.col("bond_redemption_text"),
            F.col("ecm_notes_text")
        )
    )
)
```

#Step 3

```python
from pyspark.sql import functions as F

mna_party_groups_df = (
    mna_deals_df
    .select(
        "mna_deal_id",

        F.expr("""
            transform(
                filter(
                    MNADealCompanies,
                    x -> lower(x.CompanyRoleName) = 'target'
                ),
                x -> cast(x.CompanyId as string)
            )
        """).alias("target_group_ids"),

        F.expr("""
            transform(
                filter(
                    MNADealCompanies,
                    x -> lower(x.CompanyRoleName) in (
                        'acquiror',
                        'acquiring subsidiary'
                    )
                ),
                x -> cast(x.CompanyId as string)
            )
        """).alias("acquiror_group_ids"),

        F.expr("""
            transform(
                filter(
                    MNADealCompanies,
                    x -> lower(x.CompanyRoleName) = 'target'
                ),
                x -> x.CompanyName
            )
        """).alias("target_group_names"),

        F.expr("""
            transform(
                filter(
                    MNADealCompanies,
                    x -> lower(x.CompanyRoleName) in (
                        'acquiror',
                        'acquiring subsidiary'
                    )
                ),
                x -> x.CompanyName
            )
        """).alias("acquiror_group_names")
    )
)
```

```python
financing_analysis_df = (
    financing_evidence_df
    .join(
        mna_party_groups_df,
        on="mna_deal_id",
        how="left"
    )
    .withColumn(
        "borrower_or_issuer_id_string",
        F.col("borrower_or_issuer_id").cast("string")
    )
    .withColumn(
        "issuer_relationship",
        F.when(
            F.array_contains(
                F.col("target_group_ids"),
                F.col("borrower_or_issuer_id_string")
            ),
            F.lit("Target")
        )
        .when(
            F.array_contains(
                F.col("acquiror_group_ids"),
                F.col("borrower_or_issuer_id_string")
            ),
            F.lit("Acquiror")
        )
        .otherwise(F.lit("Unresolved"))
    )
    .withColumn(
        "issuer_relationship_basis",
        F.when(
            F.col("issuer_relationship") == "Target",
            F.lit("Issuer ID matches a target recorded on the M&A deal")
        )
        .when(
            F.col("issuer_relationship") == "Acquiror",
            F.lit(
                "Issuer ID matches an acquiror or acquiring subsidiary "
                "recorded on the M&A deal"
            )
        )
        .otherwise(
            F.lit(
                "Issuer ID does not match a recorded target, acquiror "
                "or acquiring subsidiary"
            )
        )
    )
)
```

```python
financing_analysis_df = (
    financing_analysis_df
    .withColumn(
        "financing_execution_date",
        F.when(
            F.col("financing_market") == "Loan",
            F.coalesce(
                F.col("financing_funding_date"),
                F.col("financing_sign_date")
            )
        )
        .when(
            F.col("financing_market").isin("DCM", "ECM"),
            F.coalesce(
                F.col("financing_settlement_date"),
                F.col("financing_pricing_date")
            )
        )
    )
    .withColumn(
        "financing_execution_date_source",
        F.when(
            (F.col("financing_market") == "Loan")
            & F.col("financing_funding_date").isNotNull(),
            F.lit("FundingDate")
        )
        .when(
            (F.col("financing_market") == "Loan")
            & F.col("financing_sign_date").isNotNull(),
            F.lit("SignDate")
        )
        .when(
            F.col("financing_market").isin("DCM", "ECM")
            & F.col("financing_settlement_date").isNotNull(),
            F.lit("SettlementDate")
        )
        .when(
            F.col("financing_market").isin("DCM", "ECM")
            & F.col("financing_pricing_date").isNotNull(),
            F.lit("PricingDate")
        )
    )
    .withColumn(
        "financing_reference_date",
        F.coalesce(
            F.col("financing_execution_date"),
            F.col("financing_announcement_date")
        )
    )
    .withColumn(
        "financing_reference_date_source",
        F.coalesce(
            F.col("financing_execution_date_source"),
            F.lit("AnnouncementDate")
        )
    )
)
```

```python
financing_analysis_df = (
    financing_analysis_df
    .withColumn(
        "financing_timing",
        F.when(
            F.col("financing_reference_date").isNull(),
            F.lit("Financing date unavailable")
        )
        .when(
            F.col("mna_announcement_date").isNull(),
            F.lit("M&A announcement unavailable")
        )
        .when(
            F.col("financing_reference_date")
            < F.col("mna_announcement_date"),
            F.lit("Before announcement")
        )
        .when(
            F.col("mna_completion_date").isNull(),
            F.lit("Completion unavailable")
        )
        .when(
            F.col("financing_reference_date")
            <= F.col("mna_completion_date"),
            F.lit("Announcement to completion")
        )
        .otherwise(F.lit("After completion"))
    )
    .withColumn(
        "days_from_mna_announcement",
        F.datediff(
            F.col("financing_reference_date"),
            F.col("mna_announcement_date")
        )
    )
    .withColumn(
        "days_from_mna_completion",
        F.datediff(
            F.col("financing_reference_date"),
            F.col("mna_completion_date")
        )
    )
)
```

```python
loan_tranche_purpose_df = (
    loan_tranches_df
    .select(
        "mna_deal_id",
        "financing_deal_id",
        "tranche_id",

        F.array_sort(
            F.array_distinct(
                F.expr("""
                    transform(
                        LoanDealTranchesUseOfProceeds,
                        x -> x.UseName
                    )
                """)
            )
        ).alias("tranche_purpose_set")
    )
    .withColumn(
        "tranche_purpose_text",
        F.concat_ws(
            " | ",
            F.col("tranche_purpose_set")
        )
    )
)
```

```python
loan_purpose_consistency_df = (
    loan_tranche_purpose_df
    .groupBy(
        "mna_deal_id",
        "financing_deal_id"
    )
    .agg(
        F.countDistinct("tranche_id")
        .alias("purpose_tranche_count"),

        F.countDistinct("tranche_purpose_text")
        .alias("distinct_tranche_purpose_sets"),

        F.max(
            F.when(
                F.size("tranche_purpose_set") > 1,
                F.lit(1)
            ).otherwise(F.lit(0))
        ).cast("boolean")
        .alias("has_multiple_purposes_within_tranche"),

        F.collect_list(
            F.struct(
                "tranche_id",
                "tranche_purpose_set",
                "tranche_purpose_text"
            )
        ).alias("tranche_purpose_details")
    )
    .withColumn(
        "has_different_purposes_across_tranches",
        F.col("distinct_tranche_purpose_sets") > 1
    )
    .withColumn(
        "mixed_purpose_financing",
        F.col("has_multiple_purposes_within_tranche")
        | F.col("has_different_purposes_across_tranches")
    )
)
```

```python
financing_analysis_df = (
    financing_analysis_df
    .join(
        loan_purpose_consistency_df,
        on=["mna_deal_id", "financing_deal_id"],
        how="left"
    )
    .withColumn(
        "mixed_purpose_financing",
        F.when(
            F.col("financing_market") == "Loan",
            F.coalesce(
                F.col("mixed_purpose_financing"),
                F.lit(False)
            )
        )
    )
)
```

```python
financing_analysis_df = (
    financing_analysis_df
    .withColumn(
        "is_target_issuer",
        F.col("issuer_relationship") == "Target"
    )
    .withColumn(
        "is_acquiror_issuer",
        F.col("issuer_relationship") == "Acquiror"
    )
    .withColumn(
        "issuer_relationship_unresolved",
        F.col("issuer_relationship") == "Unresolved"
    )
    .withColumn(
        "financing_before_announcement",
        F.col("financing_timing") == "Before announcement"
    )
    .withColumn(
        "financing_during_transaction",
        F.col("financing_timing") == "Announcement to completion"
    )
    .withColumn(
        "financing_after_completion",
        F.col("financing_timing") == "After completion"
    )
)
```

#Creating a classification-neutral evidence-gap table

```python
#Setup:
from pyspark.sql import functions as F

KEYS = ["mna_deal_id", "financing_deal_id"]

ACQUISITION_PATTERN = (
    r"(?i)\b(lbo|mbo|leveraged buyout|acquisition financing|"
    r"finance(?:d|s|ing)? the acquisition|purchase price|cash consideration)\b"
)

REFINANCING_PATTERN = (
    r"(?i)\b(refinanc(?:e|ed|es|ing)|repay(?:ment|ing)? debt|"
    r"redeem(?:ed|ing|ption)?|exchange offer|take[- ]?out)\b"
)

DIVIDEND_PATTERN = (
    r"(?i)\b(dividend|distribution|shareholder payment|"
    r"return of capital|recapitalisation|recapitalization|dividend recap)\b"
)

BRIDGE_PATTERN = (
    r"(?i)\b(bridge|interim financing|temporary acquisition financing)\b"
)
```

```python
#Preserve loan evidence:
loan_base_df = (
    loan_tranches_df
    .select(
        *KEYS,
        "tranche_id",
        "PrimaryInstrumentTypeName",
        "MaturityDate",
        "YearsToMaturity",
        "IsBridgeToEquity",
        "IsBridgeToHighYield",
        "LoanDealTranchesUseOfProceeds",
        "LoanDealTranchesOutstandings",
        "LoanDealTranchesValue",
        "LoanDealTranchesGuarantors"
    )
)

loan_outstanding_observations_df = (
    loan_base_df
    .select(
        *KEYS,
        "tranche_id",
        F.explode_outer("LoanDealTranchesOutstandings").alias("observation")
    )
    .select(
        *KEYS,
        "tranche_id",
        F.to_date("observation.Date").alias("observation_date"),
        F.concat(
            F.coalesce(
                F.col("observation.AmountOutstanding"),
                F.array()
            ),
            F.coalesce(
                F.col("observation.AmountDrawnDown"),
                F.array()
            )
        ).alias("amounts")
    )
    .select(
        *KEYS,
        "tranche_id",
        "observation_date",
        F.explode_outer("amounts").alias("amount")
    )
    .select(
        *KEYS,
        "tranche_id",
        "observation_date",
        F.col("amount.CurrencyISOCode").alias("currency"),
        F.col("amount.Value").cast("decimal(38, 8)").alias("amount")
    )
)

loan_outstanding_summary_df = (
    loan_outstanding_observations_df
    .groupBy(*KEYS)
    .agg(
        F.min("observation_date")
        .alias("loan_first_outstanding_observation_date"),

        F.max("observation_date")
        .alias("loan_last_outstanding_observation_date"),

        F.max(
            F.when(F.col("amount") > 0, F.col("observation_date"))
        ).alias("loan_last_positive_outstanding_date"),

        F.max(
            F.when(F.col("amount") > 0, F.lit(1)).otherwise(F.lit(0))
        ).cast("boolean").alias("has_positive_loan_outstanding_observation"),

        F.collect_list(
            F.to_json(
                F.struct(
                    "tranche_id",
                    "observation_date",
                    "currency",
                    "amount"
                )
            )
        ).alias("loan_outstanding_evidence")
    )
)

loan_evidence_gap_df = (
    loan_base_df
    .withColumn(
        "loan_use_names",
        F.expr("""
            transform(
                LoanDealTranchesUseOfProceeds,
                x -> x.UseName
            )
        """)
    )
    .groupBy(*KEYS)
    .agg(
        F.countDistinct("tranche_id")
        .alias("evidence_loan_tranche_count"),

        F.min(F.to_date("MaturityDate"))
        .alias("loan_earliest_maturity_date"),

        F.max(F.to_date("MaturityDate"))
        .alias("loan_latest_maturity_date"),

        F.max(
            F.coalesce(
                F.col("IsBridgeToEquity").cast("int"),
                F.lit(0)
            )
        ).cast("boolean").alias("loan_bridge_to_equity_flag"),

        F.max(
            F.coalesce(
                F.col("IsBridgeToHighYield").cast("int"),
                F.lit(0)
            )
        ).cast("boolean").alias("loan_bridge_to_high_yield_flag"),

        F.sort_array(
            F.collect_set("PrimaryInstrumentTypeName")
        ).alias("evidence_loan_instrument_types"),

        F.sort_array(
            F.array_distinct(
                F.flatten(
                    F.collect_list("loan_use_names")
                )
            )
        ).alias("evidence_loan_use_of_proceeds"),

        F.collect_list(
            F.to_json(
                F.struct(
                    "tranche_id",
                    "PrimaryInstrumentTypeName",
                    "MaturityDate",
                    "YearsToMaturity",
                    "IsBridgeToEquity",
                    "IsBridgeToHighYield",
                    "LoanDealTranchesUseOfProceeds",
                    "LoanDealTranchesValue",
                    "LoanDealTranchesGuarantors"
                )
            )
        ).alias("loan_tranche_evidence")
    )
    .join(
        loan_outstanding_summary_df,
        KEYS,
        "left"
    )
)
```

```python
#Preserve bond evidence:
bond_evidence_gap_df = (
    bond_tranches_df
    .groupBy(*KEYS)
    .agg(
        F.countDistinct("tranche_id")
        .alias("evidence_bond_tranche_count"),

        F.min(F.to_date("MaturityDate"))
        .alias("bond_earliest_maturity_date"),

        F.max(F.to_date("MaturityDate"))
        .alias("bond_latest_maturity_date"),

        F.min(F.to_date("TakeOutDate"))
        .alias("bond_first_takeout_date"),

        F.max(F.to_date("TakeOutDate"))
        .alias("bond_last_takeout_date"),

        F.sort_array(
            F.collect_set("SecurityTypeName")
        ).alias("evidence_bond_security_types"),

        F.sort_array(
            F.collect_set("RedemptionDetails")
        ).alias("bond_redemption_details"),

        F.sort_array(
            F.collect_set("Comment")
        ).alias("bond_comments"),

        F.max(
            F.when(
                F.size("DCMDealTranchesOutstandingAmount") > 0,
                F.lit(1)
            ).otherwise(F.lit(0))
        ).cast("boolean").alias("has_bond_outstanding_amount"),

        F.max(
            F.when(
                F.col("TakeOutDate").isNotNull(),
                F.lit(1)
            ).otherwise(F.lit(0))
        ).cast("boolean").alias("has_bond_takeout_date"),

        F.collect_list(
            F.to_json(
                F.struct(
                    "tranche_id",
                    "SecurityTypeName",
                    "MaturityDate",
                    "TakeOutDate",
                    "RedemptionDetails",
                    "Comment",
                    "DCMDealTranchesOutstandingAmount",
                    "DCMDealTranchesProceeds",
                    "DCMDealTranchesValue",
                    "DCMDealTranchesGuarantors"
                )
            )
        ).alias("bond_tranche_evidence")
    )
)
```

```python
#Preserve ECM evidence:
ecm_evidence_gap_df = (
    ecm_tranches_df
    .groupBy(*KEYS)
    .agg(
        F.countDistinct("tranche_id")
        .alias("evidence_ecm_tranche_count"),

        F.sort_array(
            F.collect_set("SecurityTypeName")
        ).alias("evidence_ecm_security_types"),

        F.collect_list(
            F.to_json(
                F.struct(
                    "tranche_id",
                    "SecurityTypeName",
                    "MaturityDate",
                    "ECMDealTranchesValue",
                    "ECMDealTranchesCurrentValue"
                )
            )
        ).alias("ecm_tranche_evidence")
    )
)
```

```python
#Retain financing-level narrative evidence:
financing_narrative_evidence_df = (
    financings_df
    .select(
        *KEYS,

        F.concat_ws(
            " | ",
            F.col("Comments"),
            F.col("BookrunnerComment"),
            F.col("UseOfProceedsNote")
        ).alias("combined_financing_narrative"),

        F.to_json("DCMDealUseOfProceeds")
        .alias("dcm_use_of_proceeds_evidence"),

        F.to_json("DCMDealProceeds")
        .alias("dcm_proceeds_evidence"),

        F.to_json("ECMDealUseOfProceeds")
        .alias("ecm_use_of_proceeds_evidence"),

        F.to_json("ECMDealProceeds")
        .alias("ecm_proceeds_evidence"),

        "LoanDealNewMoneyAmount",
        "LoanDealNewMoneyAmountOverride",
        "LoanDealTotalValue",
        "LoanDealValue",
        "DCMDealTotalAmountOutstanding",
        "DCMDealTotalProceeds",
        "DCMDealTotalValue",
        "DCMDealValue",
        "ECMDealTotalValue",
        "ECMDealValue"
    )
)
```

```python
#Build the evidence-gap table:
financing_evidence_gap_df = (
    financing_analysis_df.alias("a")

    .join(
        loan_evidence_gap_df.alias("l"),
        KEYS,
        "left"
    )

    .join(
        bond_evidence_gap_df.alias("b"),
        KEYS,
        "left"
    )

    .join(
        ecm_evidence_gap_df.alias("e"),
        KEYS,
        "left"
    )

    .join(
        financing_narrative_evidence_df.alias("f"),
        KEYS,
        "left"
    )

    .withColumn(
        "combined_purpose_evidence",
        F.concat_ws(
            " | ",
            F.concat_ws(
                ", ",
                F.col("evidence_loan_use_of_proceeds")
            ),
            F.col("financing_use_of_proceeds_raw"),
            F.col("financing_use_of_proceeds_note"),
            F.col("combined_financing_narrative"),
            F.concat_ws(
                " | ",
                F.col("bond_comments")
            ),
            F.concat_ws(
                " | ",
                F.col("bond_redemption_details")
            )
        )
    )

    .withColumn(
        "existed_before_completion_status",
        F.when(
            F.col("mna_completion_date").isNull()
            | F.col("financing_reference_date").isNull(),
            F.lit("Unknown")
        )
        .when(
            F.col("financing_reference_date")
            < F.col("mna_completion_date"),
            F.lit("Pass")
        )
        .otherwise(F.lit("Fail"))
    )

    .withColumn(
        "maturity_after_completion_flag",
        F.when(
            F.col("mna_completion_date").isNull(),
            F.lit(None).cast("boolean")
        )
        .when(
            F.greatest(
                F.col("loan_latest_maturity_date"),
                F.col("bond_latest_maturity_date")
            ) > F.col("mna_completion_date"),
            F.lit(True)
        )
        .otherwise(F.lit(False))
    )

    .withColumn(
        "positive_outstanding_after_completion_flag",
        F.when(
            F.col("mna_completion_date").isNull(),
            F.lit(None).cast("boolean")
        )
        .when(
            F.col("loan_last_positive_outstanding_date")
            > F.col("mna_completion_date"),
            F.lit(True)
        )
        .otherwise(F.lit(False))
    )

    .withColumn(
        "continuity_evidence_status",
        F.when(
            F.col("positive_outstanding_after_completion_flag") == True,
            F.lit("Supported")
        )
        .when(
            F.col("loan_last_outstanding_observation_date")
            > F.col("mna_completion_date"),
            F.lit("Observed after completion but no positive balance")
        )
        .when(
            F.col("maturity_after_completion_flag") == True,
            F.lit("Maturity only - continuity unproven")
        )
        .otherwise(F.lit("Missing"))
    )

    .withColumn(
        "acquisition_funding_text_flag",
        F.col("combined_purpose_evidence")
        .rlike(ACQUISITION_PATTERN)
    )

    .withColumn(
        "refinancing_text_flag",
        F.col("combined_purpose_evidence")
        .rlike(REFINANCING_PATTERN)
    )

    .withColumn(
        "dividend_text_flag",
        F.col("combined_purpose_evidence")
        .rlike(DIVIDEND_PATTERN)
    )

    .withColumn(
        "bridge_text_flag",
        F.col("combined_purpose_evidence")
        .rlike(BRIDGE_PATTERN)
    )

    .withColumn(
        "bridge_evidence_flag",
        F.coalesce(
            F.col("loan_bridge_to_equity_flag"),
            F.lit(False)
        )
        | F.coalesce(
            F.col("loan_bridge_to_high_yield_flag"),
            F.lit(False)
        )
        | F.coalesce(
            F.col("bridge_text_flag"),
            F.lit(False)
        )
    )

    .withColumn(
        "target_relationship_status",
        F.when(
            F.col("issuer_relationship") == "Target",
            F.lit("Pass")
        )
        .when(
            F.col("issuer_relationship") == "Acquiror",
            F.lit("Fail")
        )
        .otherwise(F.lit("Unknown"))
    )

    .withColumn(
        "acquisition_structure_relationship_status",
        F.when(
            F.col("issuer_relationship") == "Acquiror",
            F.lit("Partially supported")
        )
        .when(
            F.col("issuer_relationship") == "Target",
            F.lit("Target issuer")
        )
        .otherwise(F.lit("Unknown - acquisition vehicle review required"))
    )

    .withColumn(
        "predecessor_financing_status",
        F.when(
            F.col("refinancing_text_flag")
            | F.col("has_bond_takeout_date"),
            F.lit("Required - predecessor not yet structured")
        )
        .otherwise(F.lit("Not indicated by current evidence"))
    )

    .withColumn(
        "direct_evidence_gap",
        F.concat_ws(
            "; ",
            F.when(
                ~F.coalesce(
                    F.col("acquisition_funding_text_flag"),
                    F.lit(False)
                ),
                F.lit("Acquisition-funding purpose not proven")
            ),
            F.when(
                F.col("issuer_relationship") == "Unresolved",
                F.lit("Issuer/acquisition-vehicle relationship unresolved")
            ),
            F.when(
                F.col("refinancing_text_flag")
                & ~F.col("bridge_evidence_flag"),
                F.lit("Refinancing identified but bridge predecessor not proven")
            )
        )
    )

    .withColumn(
        "carried_evidence_gap",
        F.concat_ws(
            "; ",
            F.when(
                F.col("existed_before_completion_status") != "Pass",
                F.lit("Existence before completion not proven")
            ),
            F.when(
                F.col("target_relationship_status") != "Pass",
                F.lit("Target-group relationship not proven")
            ),
            F.when(
                F.col("continuity_evidence_status") != "Supported",
                F.lit("Positive post-completion outstanding evidence missing")
            ),
            F.when(
                F.col("acquisition_funding_text_flag"),
                F.lit("Current acquisition-funding purpose must be excluded")
            )
        )
    )

    .withColumn(
        "subsequent_evidence_gap",
        F.concat_ws(
            "; ",
            F.when(
                F.col("financing_timing") != "After completion",
                F.lit("Post-completion timing not established")
            ),
            F.when(
                F.col("refinancing_text_flag")
                & (
                    F.col("predecessor_financing_status")
                    == "Required - predecessor not yet structured"
                ),
                F.lit("Predecessor financing not identified")
            ),
            F.when(
                F.col("dividend_text_flag"),
                F.lit("Dividend amount and source-and-uses evidence required")
            ),
            F.when(
                ~F.coalesce(F.col("refinancing_text_flag"), F.lit(False))
                & ~F.coalesce(F.col("dividend_text_flag"), F.lit(False)),
                F.lit("Neither refinancing nor dividend-recap purpose proven")
            )
        )
    )

    .withColumn(
        "manual_review_priority",
        F.when(
            F.col("issuer_relationship") == "Unresolved",
            F.lit(1)
        )
        .when(
            F.col("financing_timing") == "After completion",
            F.lit(1)
        )
        .when(
            F.col("refinancing_text_flag")
            | F.col("dividend_text_flag"),
            F.lit(1)
        )
        .when(
            F.col("mixed_purpose_financing") == True,
            F.lit(2)
        )
        .when(
            F.col("existed_before_completion_status") == "Pass",
            F.lit(2)
        )
        .otherwise(F.lit(3))
    )

    .withColumn(
        "manual_review_reason",
        F.concat_ws(
            "; ",
            F.when(
                F.col("issuer_relationship") == "Unresolved",
                F.lit("Unresolved issuer relationship")
            ),
            F.when(
                F.col("financing_timing") == "After completion",
                F.lit("Post-completion financing")
            ),
            F.when(
                F.col("refinancing_text_flag"),
                F.lit("Refinancing or redemption evidence")
            ),
            F.when(
                F.col("dividend_text_flag"),
                F.lit("Possible dividend recapitalisation")
            ),
            F.when(
                F.col("mixed_purpose_financing") == True,
                F.lit("Mixed-purpose financing package")
            ),
            F.when(
                F.col("continuity_evidence_status")
                == "Maturity only - continuity unproven",
                F.lit("Possible Carried debt without continuity proof")
            )
        )
    )
)
```

```python
#Select the final review table:
financing_evidence_gap_df = (
    financing_evidence_gap_df
    .select(
        *KEYS,

        # Existing identification
        "target_name",
        "acquiror_name",
        "borrower_or_issuer_name",
        "financing_market",
        "issuer_relationship",
        "issuer_relationship_basis",
        "financing_timing",
        "reviewed_by_will",

        # Existing dates
        "mna_announcement_date",
        "mna_completion_date",
        "financing_announcement_date",
        "financing_reference_date",
        "financing_reference_date_source",

        # Existing financing evidence
        "loan_instrument_types",
        "bond_security_types",
        "ecm_security_types",
        "loan_use_of_proceeds",
        "financing_use_of_proceeds_raw",
        "financing_use_of_proceeds_note",
        "mixed_purpose_financing",

        # Additional dates and amounts
        "loan_earliest_maturity_date",
        "loan_latest_maturity_date",
        "bond_earliest_maturity_date",
        "bond_latest_maturity_date",
        "bond_first_takeout_date",
        "bond_last_takeout_date",
        "loan_first_outstanding_observation_date",
        "loan_last_outstanding_observation_date",
        "loan_last_positive_outstanding_date",

        # Evidence indicators
        "existed_before_completion_status",
        "target_relationship_status",
        "acquisition_structure_relationship_status",
        "maturity_after_completion_flag",
        "positive_outstanding_after_completion_flag",
        "continuity_evidence_status",
        "acquisition_funding_text_flag",
        "refinancing_text_flag",
        "dividend_text_flag",
        "bridge_evidence_flag",
        "predecessor_financing_status",

        # Gaps
        "direct_evidence_gap",
        "carried_evidence_gap",
        "subsequent_evidence_gap",
        "manual_review_priority",
        "manual_review_reason",

        # Source evidence
        "combined_purpose_evidence",
        "bond_comments",
        "bond_redemption_details",
        "loan_tranche_evidence",
        "loan_outstanding_evidence",
        "bond_tranche_evidence",
        "ecm_tranche_evidence"
    )
)
```
